```python
from time import time
import plotly.graph_objects as go
from IPython.display import display, clear_output

# Create an empty figure
fig = go.Figure()

# Assuming mvco DataFrame is already defined
for i in range(len(mvco.columns)):
    if i == 0: continue # skip the date time column
    if i > 4 and i < 43: continue # skip whatever is after ocean variables (wave) and before QC (quality control) variables
    trace = go.Scatter(
        x=mvco.index,
        y=mvco[mvco.columns[i]],
        name=str(mvco.columns[i]),
        line=dict(color=unique_colors[i]),
        connectgaps=False,
        opacity=0.5
    )
    fig.add_trace(trace)
    
    # Clear the previous output and display the updated figure
    clear_output(wait=True)
    display(fig)
    

# Final display of the complete figure
clear_output(wait=True)
display(fig)
```
